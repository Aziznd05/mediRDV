```python
df = df.drop_duplicates().reset_index(drop=True)
print("Dimensions après suppression :", df.shape)
```

```python
df["date_prise_parsee"] = pd.to_datetime(df["date_prise"], errors="coerce", dayfirst=False)
df["date_rdv_parsee"]   = pd.to_datetime(df["date_rdv"],   errors="coerce", dayfirst=False)

# Délai en jours
df["delai_jours"] = (df["date_rdv_parsee"] - df["date_prise_parsee"]).dt.days

print("RDV pris après la date du RDV (délai < 0) :", (df["delai_jours"] < 0).sum())
print("RDV à plus de 90 jours :", (df["delai_jours"] > 90).sum())
print("\nDélai min :", df["delai_jours"].min())
print("Délai max :", df["delai_jours"].max())
print("Délai moyen :", round(df["delai_jours"].mean(), 1))
```

```python
# Trouver les dates au format JJ/MM/AAAA
masque = df["date_rdv"].astype(str).str.contains("/", na=False)
print("Dates au format JJ/MM/AAAA :", masque.sum())
print("\nExemples :")
print(df[masque][["id_rdv", "date_prise", "date_rdv"]].head(10))
```

```python
def parser_date(val):
    if pd.isna(val):
        return pd.NaT
    val = str(val).strip()
    for fmt in ("%Y-%m-%d", "%d/%m/%Y"):
        try:
            return pd.to_datetime(val, format=fmt)
        except ValueError:
            continue
    return pd.NaT

df["date_prise_parsee"] = df["date_prise"].apply(parser_date)
df["date_rdv_parsee"]   = df["date_rdv"].apply(parser_date)

print("date_prise non parsées :", df["date_prise_parsee"].isna().sum())
print("date_rdv non parsées   :", df["date_rdv_parsee"].isna().sum())

# Recalcul du délai
df["delai_jours"] = (df["date_rdv_parsee"] - df["date_prise_parsee"]).dt.days

print("\n--- RG-04 : Délai de prise ---")
print("RDV avant la date de prise :", (df["delai_jours"] < 0).sum())
print("RDV à plus de 90 jours     :", (df["delai_jours"] > 90).sum())
print("Délai min :", df["delai_jours"].min())
print("Délai max :", df["delai_jours"].max())
print("Délai moyen :", round(df["delai_jours"].mean(), 1))
```

```python
# Doublons d'id_rdv
ids_dupliques = df[df.duplicated(subset="id_rdv", keep=False)].sort_values("id_rdv")
print("Lignes concernées :", len(ids_dupliques))
print("Id_rdv uniques concernés :", ids_dupliques["id_rdv"].nunique())

# Doublons strictement identiques (toutes les colonnes)
doublons_exacts = df[df.duplicated(keep=False)]
print("\nLignes strictement identiques :", len(doublons_exacts))
```
